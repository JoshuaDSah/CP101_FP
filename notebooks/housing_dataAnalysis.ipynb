```python
import geopandas as gpd
import matplotlib.pyplot as plt
import os
import pandas as pd
import re
import numpy as np
```

```python
#Data Loading
sc_TODexistingRes = gpd.read_file("/content/drive/MyDrive/CYPLAN101_FinalProject/Data/Cleaned/sc_TOD_existing.geojson")
sc_TODcapacityRes = gpd.read_file("/content/drive/MyDrive/CYPLAN101_FinalProject/Data/Cleaned/sc_TOD_capacity.geojson")


sf_TODexistingRes = gpd.read_file("/content/drive/MyDrive/CYPLAN101_FinalProject/Data/Cleaned/sf_TODexistingRes.geojson")
sf_TODcapacityRes = gpd.read_file("/content/drive/MyDrive/CYPLAN101_FinalProject/Data/Cleaned/sf_TODcapacity_parcels.geojson")
```

```python
def plotting(x, y, col, title):
  combinedExisting = np.concatenate([x[col].dropna(), y[col].dropna()])
  bins = np.linspace(combinedExisting.min(), combinedExisting.max(), 51)

  plt.figure(figsize=(8, 5))
  plt.hist(x[col], bins=bins, alpha=0.35, color="tab:blue",
         edgecolor="tab:blue", linewidth=1.2, log=True, label="Santa Clara County")
  plt.hist(y[col], bins=bins, alpha=0.35, color="tab:red",
         edgecolor="tab:red", linewidth=1.2, log=True, label="San Francisco County")

  plt.title(title)
  plt.xlabel("Total Capacity (Units)")
  plt.ylabel("Parcel Count (Log Scale)")
  plt.legend()
  plt.grid(True, which="both", linestyle="--", linewidth=0.5)
  plt.show()
```

```python
plotting(sc_TODexistingRes,
         sf_TODexistingRes,
         "estimated_parcel_capacity",
         "Distribution of Current Site Capacity for Units in SB79 affected zones (Log Scale Y-Axis)")
```

```python
plotting(sc_TODcapacityRes,
         sf_TODcapacityRes,
         "total_capacity",
         "Distribution of Planned Increase in Site Capacity for Units in SB79 affected zones (Log Scale Y-Axis)")
```
